# Python Logging Fundamentals

Learn structured application logging with levels, handlers, formatters, and exception logging.


## 1. Basic logging
```python
import logging

logging.basicConfig(level=logging.INFO, format="%(levelname)s:%(message)s")
logging.info("Application started")
logging.warning("Example warning")
```

## 2. Log levels
Use DEBUG for diagnostic detail, INFO for normal operation, WARNING for unexpected but recoverable conditions, ERROR for failures, and CRITICAL for severe failures.

## 3. Exception logging
```python
logger = logging.getLogger(__name__)
try:
    result = 10 / 0
except ZeroDivisionError:
    logger.exception("Calculation failed")
```

## 4. Handlers and formatters
Separate where logs go from how they are formatted. Avoid hard-coded credentials or machine-specific paths.
