# Regular Expressions

Learn pattern matching, extraction, validation, groups, and substitution with Python's `re` module.


## 1. Matching text
```python
import re

text = "Order IDs: ORD-101, ORD-205"
ids = re.findall(r"ORD-\d+", text)
print(ids)
```

## 2. Groups
Use capture groups to extract structured information such as country codes, dates, or product IDs.

## 3. Validation
```python
email_pattern = r"^[^@\s]+@[^@\s]+\.[^@\s]+$"
print(bool(re.fullmatch(email_pattern, "student@example.com")))
```

## 4. Substitution
Use `re.sub()` to normalize whitespace or redact sensitive-looking values.
