# Modules and Packages

Learn how Python modules, packages, imports, and environments organize reusable code.


## 1. Modules and imports
A module is a `.py` file containing reusable code.

```python
import math
from pathlib import Path

print(math.sqrt(81))
print(Path.cwd())
```

## 2. Creating a module
Create `math_utils.py` with functions, then import it from another script or notebook.

```python
# math_utils.py
def add(a, b):
    return a + b

def percentage(value, total):
    if total == 0:
        raise ValueError("total cannot be zero")
    return value / total * 100
```

## 3. Packages
A package groups related modules in a directory. Use clear names and keep imports explicit.

## 4. Virtual environments and dependencies
Create an isolated environment with `python -m venv .venv`, activate it, and install dependencies from `requirements.txt`.
