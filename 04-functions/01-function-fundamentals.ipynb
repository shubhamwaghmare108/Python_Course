# Function Fundamentals

Learn how to define, call, parameterize, document, and return values from Python functions.


## 1. Defining a function

```python
def greet(name):
    return f"Hello, {name}!"

print(greet("Shubham"))
```


## 2. Parameters and arguments

Use positional and keyword arguments. Give parameters meaningful names.


## 3. Return values

A function should return a result when the caller needs to use that result later.


## 4. Default arguments

```python
def calculate_discount(price, rate=0.10):
    return price * (1 - rate)
```


## 5. Scope

Local variables belong to the function. Avoid unnecessary global state.


## 6. Docstrings and type hints

```python
def add(a: float, b: float) -> float:
    """Return the sum of two numbers."""
    return a + b
```
