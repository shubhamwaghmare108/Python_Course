# Conditional Statements

Learn `if`, `elif`, `else`, nested conditions, comparison operators, truthiness, and short-circuit evaluation.


## 1. Basic `if`

```python
age = 20
if age >= 18:
    print("Adult")
```


## 2. `if` / `elif` / `else`

```python
score = 72
if score >= 90:
    grade = "A"
elif score >= 75:
    grade = "B"
elif score >= 60:
    grade = "C"
else:
    grade = "D"
print(grade)
```


## 3. Nested conditions

Use nesting only when it makes the decision logic clearer.


## 4. Truthy and falsy values

Empty strings, empty collections, `0`, `None`, and `False` are falsy. Most other objects are truthy.


## 5. Short-circuit evaluation

`and` stops at the first falsy value; `or` stops at the first truthy value.
