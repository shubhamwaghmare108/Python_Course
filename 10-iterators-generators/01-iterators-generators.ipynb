# Iterators and Generators

Understand Python's iteration protocol and build memory-efficient generators.


## 1. Iterable vs iterator
```python
numbers = [10, 20, 30]
iterator = iter(numbers)
print(next(iterator))
print(next(iterator))
```

## 2. Generator functions
```python
def squares(limit):
    for number in range(limit + 1):
        yield number ** 2

print(list(squares(5)))
```

## 3. Generator expressions
Generator expressions compute values lazily and are useful for large sequences.

## 4. When generators help
Prefer generators when the full result does not need to be held in memory at once.
