# Decorators

Learn closures and decorators for wrapping functions without changing their core implementation.


## 1. Functions are objects
Functions can be stored in variables, passed to other functions, and returned from functions.

## 2. A simple decorator
```python
def log_call(func):
    def wrapper(*args, **kwargs):
        print(f"Calling {func.__name__}")
        return func(*args, **kwargs)
    return wrapper

@log_call
def add(a, b):
    return a + b

print(add(2, 3))
```

## 3. Preserving metadata
Use `functools.wraps` so the wrapped function retains useful metadata such as its name and docstring.

## 4. Practical patterns
Decorators are useful for logging, timing, validation, authorization, and caching.
