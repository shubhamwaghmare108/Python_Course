## Practice
Combine an employee table with a department table using a left join, then concatenate two monthly employee snapshots.


```python
import pandas as pd

employees = pd.DataFrame({'emp_id':[1,2,3],'name':['Asha','Ravi','Neha'],'dept_id':[10,20,10]})
departments = pd.DataFrame({'dept_id':[10,20],'department':['Data','IT']})
result = employees.merge(departments, on='dept_id', how='left', validate='many_to_one')
print(result)
jan = employees.iloc[:2]
feb = employees.iloc[1:]
print(pd.concat([jan, feb], ignore_index=True))

```
