## Practice
Create a score Series, calculate the mean, find the highest score, filter passing students, and handle one missing value.


```python
import pandas as pd

scores = pd.Series([72, 85, 91, 68], index=['Asha','Ravi','Neha','Vikram'], name='score')
print(scores)
print(scores.loc['Ravi'])
print(scores.iloc[0])
print(scores.mean())

```
