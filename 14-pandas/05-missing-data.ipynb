## Practice
Create a DataFrame with missing age and salary values. Report missing counts, fill numeric values with medians, and explain the choice.


```python
import pandas as pd

employees = pd.DataFrame({'name':['Asha','Ravi','Neha','Vikram'],'age':[25,None,31,29],'salary':[55000,62000,None,70000]})
print(employees.isna().sum())
employees['age'] = employees['age'].fillna(employees['age'].median())
employees['salary'] = employees['salary'].fillna(employees['salary'].median())
print(employees)

```
