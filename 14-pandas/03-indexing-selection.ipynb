# Pandas Indexing and Selection

Select rows and columns safely using labels and positions.

## Topics
- column selection
- `.loc` label selection
- `.iloc` positional selection
- `.at` and `.iat`
- selecting rows and subsets

## Practice
Given a student DataFrame, select one column, multiple columns, a row by label, and a rectangular subset by position.


```python
import pandas as pd

df = pd.DataFrame({'name':['Asha','Ravi','Neha'],'score':[72,85,91],'city':['Pune','Mumbai','Nagpur']}, index=['S1','S2','S3'])
print(df['score'])
print(df[['name','score']])
print(df.loc['S2'])
print(df.iloc[:2, :2])

```
