Turn business conditions into boolean masks and queries.

## Topics
- boolean masks
- multiple conditions with `&` and `|`
- `.isin()` and `.between()`
- `.query()`
- sorting filtered results

## Practice
Find employees in Data or IT whose salary is at least 55000, then sort by salary descending.


```python
import pandas as pd

df = pd.DataFrame({'name':['Asha','Ravi','Neha','Vikram'],'department':['Data','IT','HR','Data'],'salary':[55000,62000,48000,70000]})
mask = df['department'].isin(['Data','IT']) & df['salary'].ge(55000)
print(df.loc[mask].sort_values('salary', ascending=False))

```
